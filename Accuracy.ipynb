```python
import matplotlib.pyplot as plt

# Define model names and accuracy values
models = ["YOLOv8", "Faster R-CNN", "EfficientDet", "SSD", "YOLOv11"]
accuracy = [52.8, 49.1, 51.3, 44.7, 69.5]  # YOLOv11 has the highest accuracy

# Define colors, highlighting YOLOv11
colors = ["gray", "gray", "gray", "gray", "blue"]

# Create bar chart
plt.figure(figsize=(8, 5))
plt.bar(models, accuracy, color=colors, alpha=0.8)

# Add labels and title
plt.xlabel("Model")
plt.ylabel("Accuracy (%)")
plt.title("Comparison of YOLOv11 Accuracy with Existing Models")

# Annotate bars with accuracy values
for i, acc in enumerate(accuracy):
    plt.text(i, acc + 1, f"{acc}%", ha="center", fontsize=10)

# Adjust y-axis limits
plt.ylim(40, 75)

# Show the chart
plt.show()
```
