```python
from ultralytics import YOLO

# Load YOLOv11n model (lightweight, fast)
model = YOLO('yolo11n.pt')
```

```python
results = model.train(
    data='C:/Users/Abhishek/Desktop/Object Detection model/BasketballHoop/data.yaml',
    epochs=10,
    imgsz=640,
    batch=16,
    device='cpu'  # <-- set to 'cpu'
)
```

```python
from ultralytics import YOLO

model = YOLO('runs/detect/train2/weights/best.pt')
# Run inference on an image
results = model('image.jpg')

# Show each result
for r in results:
    r.show()
```
